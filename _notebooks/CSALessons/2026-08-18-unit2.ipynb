# 4.2 for Loops

## for Loops vs. while Loops
* Both are iterative statements that repeatedly execute a code block as long as a condition is met.
* **while Loops:** Use when the total number of iterations is **unknown**.
* **for Loops:** Use when the exact number of iterations is **known**.

### Code Comparison

```java
// while loop approach
int i = 0;
while (i < 5) {
    System.out.println(i);
    i++;
}

// for loop approach
for (int i = 0; i < 5; i++) {
    System.out.println(i);
}

````python
int sum = 0;
int sum_all = 0;

for (int i = 0; i < 10; i++) {
    if (i % 2 == 0) {
        sum += i;     // Adds even numbers only (0 + 2 + 4 + 6 + 8)
    }
    sum_all += i;     // Adds every number (0 through 9)
}

System.out.println(sum);     // Output: 20
System.out.println(sum_all); // Output: 45
````

````python
public class VowelCounter {
    public static void main(String[] args) {
        String word = "iterate";
        int vowelCount = 0;

        for (int i = 0; i < word.length(); i++) {
            String ch = word.substring(i, i + 1);
            if (ch.equalsIgnoreCase("a") || ch.equalsIgnoreCase("e") || 
                ch.equalsIgnoreCase("i") || ch.equalsIgnoreCase("o") || 
                ch.equalsIgnoreCase("u")) {
                vowelCount++;
            }
        }

        System.out.println("Total vowels: " + vowelCount); // Output: 4
    }
}
````

## Key Java String Tools
* **`String.substring(start, end)`:** Retrieves a portion of a string from index `start` up to (but not including) `end`.
* **`String.equals(otherString)`:** Compares content equality between two string objects (**never** use `==`).
* **`String.length()`:** Returns the total number of characters in the string.
* **`for` Loops:** Used to iterate through individual characters or substring "windows" of a string.

---

## Substring Search: The "Window" Method

To find a target substring within a main string, create a moving "window" equal to the length of the substring and shift it through the main string index-by-index.

### Buggy Code Example & Analysis

```java
public class StringFinder {
    public static void main(String[] args) {
        String word = "iterate";
        String sub = "xxx";
        boolean found = false;

        // BUG: word.length() causes StringIndexOutOfBoundsException
        for (int i = 0; i < word.length(); i++) { 
            String portion = word.substring(i, i + sub.length());
            if (sub.length() > portion.length()) break;
            if (portion.equals(sub)) {
                found = true;
                break;
            }
        }

        if (found)
            System.out.println("substring is found within string!");
