````python
def calculate_grade(score1, score2, score3):
    """
    Calculate letter grade from three test scores
    
    Args:
        score1, score2, score3: Test scores (integers)
    
    Returns:
        grade: Letter grade (string)
    """
    average_score = (score1 + score2 + score3) / 3
    
    if average_score >= 90:
        grade = 'A'
    elif average_score >= 80:
        grade = 'B'
    elif average_score >= 70:
        grade = 'C'
    elif average_score >= 60:
        grade = 'D'
    else:
        grade = 'F'

    return grade

print("Test 1:", calculate_grade(95, 92, 88))  # Should be 'A'
print("Test 2:", calculate_grade(85, 80, 82))  # Should be 'B'
print("Test 3:", calculate_grade(75, 70, 72))  # Should be 'C'
print("Test 4:", calculate_grade(65, 60, 62))  # Should be 'D'
print("Test 5:", calculate_grade(55, 50, 52))  # Should be 'F'
````

## 1.2 Variables and Data Types

<strong>Variable: </strong> named storage location in memory for data. The more specific the name, the better

* naming conventions... lowercase first letter then subsequent words are capitalized (e.g. favoriteFood)

<strong>Declaration: </strong> creating a variable with a type.

<strong>Initalization: </strong> giving the variable a value at creation. (ex. initial_price = 5)

<strong>Assignment: </strong> giving a value later, after declaration

<em> The difference between these three is that declaration creates the variable (int number), assignment stores the value in the variable (number = 10), while intialization combines the two (int age = 16). </em>

<strong>Primitive Type: </strong> basic data types built into Java (int, double, boolean, char), they store the actual value and cannot be `null`

<strong>Reference Type: </strong> points to objects; more complex (Strong, Scanner). they can be `null`. 

<em> Primitive types don't have methods while Reference types are objects and will have methods associated with them </em>

<strong>Final Variables: </strong> These will <em> never </em> change. just put final at the beginning and give it a value. `final int MAX_STUDENTS = 30;` USE ALL CAPS (e.g. BIRTH_YEAR)

### Scanner Methods Overview

| Method | Returns | Example |
|--------|---------|---------|
| `nextInt()` | integer | reads `42` |
| `nextDouble()` | decimal | reads `3.14` |
| `next()` | single word | reads `"Hello"` (stops at space) |
| `nextLine()` | entire line | reads `"Hello World"` (includes spaces) |

### Basic Input Example

```java
import java.util.Scanner;

public class InputDemo {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        
        // Reading a String (entire line)
        System.out.print("Enter your name: ");
        String name = sc.nextLine();
        System.out.println("Hello, " + name + "!");
        
        // Reading an integer
        System.out.print("Enter your age: ");
        int age = sc.nextInt();
        System.out.println("You entered: " + age);
        
        // Reading a double
        System.out.print("Enter a decimal: ");
        double value = sc.nextDouble();
        System.out.println("Double that is: " + (value * 2));
        
        sc.close();  // Always close to prevent resource leaks
    }
}
```

### ⚠️ Critical Pitfall: nextInt() / nextLine() Problem

**The Issue:**
- `nextInt()`, `nextDouble()`, and `next()` read only their token
- They **do NOT consume the newline character (`\n`)** left in the input buffer
- If you then call `nextLine()`, it immediately reads the leftover `\n` and returns an empty string

**Visual Example:**

Input Buffer: [2][5][\n][R][e][m][a][i][n][i][n][g]...

↑

nextInt() reads "25" but leaves [\n] behind

nextLine() immediately consumes [\n] and returns ""

**The Solution: Use an extra `sc.nextLine()` to consume the newline**

```java
Scanner sc = new Scanner(System.in);

System.out.print("Enter your age: ");
int age = sc.nextInt();

// FIX: Consume the leftover newline character
sc.nextLine();

System.out.print("Enter your full name: ");
String fullName = sc.nextLine();  // Now waits for input correctly

System.out.println("Hello, " + fullName + "! You are " + age + " years old.");
sc.close();
```

**Task:** Write an explanation distinguishing between:
- Assignment operator (=)
- Comparison operator (==)

Include code examples showing proper usage of each. Explain a common bug that occurs when these are confused.

**Solution:**

#### Assignment Operator (=)
- **Purpose:** Stores a value into a variable
- **Usage:** Left side is the variable being assigned, right side is the value or expression
- **Example:**
```java
int x = 5;      // Assigns 5 to x
x = x + 3;      // Evaluates x+3, then assigns result (8) back to x
String name = "John";  // Assigns String to name
```

#### Comparison Operator (==)
- **Purpose:** Compares two values and returns true/false
- **Usage:** Used in if statements and conditions
- **Example:**
```java
if (x == 5) {   // Checks if x equals 5
    System.out.println("x is 5");
}

if (name == "John") {  // Compares values
    System.out.println("Name is John");
}
```

#### Common Bug: Confusing = and ==

```java
// WRONG: Using = in an if statement
if (x = 5) {  // This ASSIGNS 5 to x, doesn't compare!
    // This will cause a type error
}

// CORRECT: Using == to compare
if (x == 5) {  // This COMPARES x to 5
    System.out.println("x equals 5");
}
```

**Why this matters:** Using `=` instead of `==` in a condition assigns a value instead of comparing, which causes unexpected behavior and compiler errors.

## 1.5 Casting and Range of Variables

### Casting
| Type | Direction | Loss? | Example |
|------|-----------|-------|---------|
| int → double | Widening (automatic) | No | `double x = 5;` |
| double → int | Narrowing (explicit) | Yes | `int x = (int)5.9;` → `5` |

### Integer Division & Modulus
- **Integer division:** Both operands int → result is int with truncation
  - `7 / 2 = 3`
  - `7 % 2 = 1`
- **Property:** `x == (x / y) * y + (x % y)`

### Range of int
| Value | Amount |
|-------|--------|
| `Integer.MAX_VALUE` | `2147483647` |
| `Integer.MIN_VALUE` | `-2147483648` |
| **Overflow behavior** | Wraps around within 32-bit range |

### Rounding (per AP CSA CED)
- **Non-negative:** `(int)(x + 0.5)`
- **Negative:** `(int)(x - 0.5)`

## Casting in Java

### Key Concept: Truncation vs Promotion

```java
int a = 7, b = 2;
System.out.println(a / b);          // 3    (int division)
System.out.println((double)a / b);  // 3.5  (promoted to double first)

double d = 5.9;
System.out.println((int)d);         // 5    (truncate toward 0)
System.out.println((int)(d + 0.5)); // 6    (round to nearest)
```

### Overflow Example

```java
int max = Integer.MAX_VALUE;  // 2147483647
System.out.println(max + 1);  // -2147483648  (overflow wraps)
```

### Promotion Rules

```java
int a = 3, b = 2;
double x = a / b;          // 1.0   (int division first, then widen)
double y = (double)a / b;  // 1.5   (cast a to double first)
```

## 1.6 Compound Assignment Operators


## Quick Reference: Compound Assignment Operators

| Operator | Full Form | Purpose |
|----------|-----------|---------|
| `+=` | `x = x + y` | Add; accumulate; concatenate strings |
| `-=` | `x = x - y` | Subtract; reduce |
| `*=` | `x = x * y` | Multiply; scale; growth |
| `/=` | `x = x / y` | Divide; scale down (⚠️ integer division if both int) |
| `%=` | `x = x % y` | Remainder; wrap-around logic |
| `++` | `x = x + 1` | Increment (post-form only on AP exam) |
| `--` | `x = x - 1` | Decrement (post-form only on AP exam) |

---

## Why Use Them?

**Before:**
```java
score = score + 10;
health = health - 5;
count = count + 1;
```

**After:**
```java
score += 10;
health -= 5;
count++;
```

Cleaner, faster to write, and easier to read!

---

## Examples

```java
public class CompoundOpsDemo {
    public static void main(String[] args) {
        int score = 100;
        double average = 85.5;
        
        score += 50;       // score = 150
        score -= 25;       // score = 125
        score *= 2;        // score = 250
        score /= 5;        // score = 50
        score %= 15;       // score = 5
        average += 4.5;    // average = 90.0
        
        int count = 0;
        count++;           // count = 1
        count--;           // count = 0
        
        System.out.println(score);   // 5
        System.out.println(count);   // 0
    }
}
```

## Part 3: Preconditions and Postconditions

### Definitions

- **Preconditions:** What MUST be true BEFORE the method is called
- **Postconditions:** What WILL be true AFTER the method completes successfully

These create a **contract** between the method and its callers.

### Example: Bank Withdrawal

```java
/**
 * Withdraws money from the bank account.
 * 
 * Preconditions:
 * - amount must be positive
 * - amount must not exceed current balance
 * - account must not be frozen
 * 
 * Postconditions:
 * - balance is reduced by the withdrawal amount
 * - transaction is recorded in account history
 * - returns true if successful, false otherwise
 *
 * @param amount the amount to withdraw (must be positive)
 * @return true if withdrawal successful, false otherwise
 */
public boolean withdraw(double amount) {
    double balance = 500.0;
    boolean isFrozen = false;
    
    // Preconditions check
    if (amount <= 0 || amount > balance || isFrozen) {
        return false;
    }
    
    // Perform withdrawal
    balance -= amount;
    System.out.println("Withdrew: $" + amount + ", New balance: $" + balance);
    
    return true;  // Postcondition met
}
```

## Part 5: Documentation Best Practices

### ✅ DO:

## 1.9 Method Signatures 

## Quick Reference

| Concept | Definition |
|---------|-----------|
| **Signature (AP scope)** | `methodName(type1, type2, ...)` — only name + parameter types/order |
| **NOT in signature** | Return type, parameter names, access modifiers, throws clauses |
| **Overloading** | Same name, different parameter lists (type, arity, or order) |
| **Widening** | int → double allowed at call site (automatic promotion) |
| **Narrowing** | double → int requires explicit cast |

---

## What IS a Method Signature?

```java
public static int add(int a, int b) {
    return a + b;
}

// SIGNATURE (AP scope): add(int, int)
// NOT part of signature:
//   - public (access modifier)
//   - static (modifier)
//   - int (return type)
//   - a, b (parameter names)
```

**Only the method name and parameter types (in order) define the signature.**

---

## What is NOT in the Signature?

❌ Return type — Cannot overload by changing only return type  
❌ Parameter names — `add(int x, int y)` and `add(int a, int b)` are the SAME signature  
❌ Access modifiers — `public` vs `private` don't change signature  
❌ `throws` clauses — Don't affect the signature  

---

## Method Overloading

**Overloading** = same method name, different parameter lists

### Valid Overloads

```java
public class Calculator {
    // Signature: add(int, int)
    public static int add(int a, int b) {
        return a + b;
    }
    
    // Signature: add(double, double) — different types
    public static double add(double a, double b) {
        return a + b;
    }
    
    // Signature: add(int) — different arity (count)
    public static int add(int a) {
        return a + 1;
    }
    
    // Signature: add(int, double) — different order/types
    public static double add(int a, double b) {
        return a + b;
    }
}
```

### NOT Valid Overloads

```java
// ❌ ERROR: Can't overload by return type alone
public static int add(int a, int b) { return a + b; }
public static double add(int a, int b) { return a + b; }  // COMPILE ERROR
```

---

## Overload Resolution (How Java Picks)

Java chooses the **most specific applicable method**.

```java
static void show(int x)    { System.out.println("int"); }
static void show(double x) { System.out.println("double"); }
static void show(long x)   { System.out.println("long"); }

public static void main(String[] args) {
    show(5);      // int literal → show(int)
    show(5L);     // long literal → show(long)
    show(5.0);    // double literal → show(double)
    
    int i = 7;
    show(i);      // variable type int → show(int)
}
```

**Rules:**
- Exact match is preferred
- Widening (int → double) is allowed
- Narrowing (double → int) requires explicit cast
- Avoid ambiguous designs

---

## Parameter Passing

**Primitives are passed by value** — changing parameter doesn't affect original variable.

```java
public static void bump(int x) {
    x = x + 1;  // Only changes local x
}

public static void main(String[] args) {
    int a = 5;
    bump(a);
    System.out.println(a);  // Still 5, NOT 6
}
```

---

## Anatomy of a Method Header

```java
public static int sum(int a, int b) {
    //     ↑      ↑    ↑   ↑ ↑    ↑
    //  modifier return method | parameters
    //           type   name  signature
}

// SIGNATURE: sum(int, int)
```

| Part | Scope | Notes |
|------|-------|-------|
| `public` | Modifier | Controls access |
| `static` | Modifier | Class method vs instance |
| `int` | Return type | NOT part of signature |
| `sum` | Method name | Part of signature |
| `(int a, int b)` | Parameters | Types and order in signature |

---

## Method Overloading Checklist

✅ Same method name  
✅ Different parameter types OR different count OR different order  
❌ Return type alone doesn't create overload  
✅ Document which overload is chosen

## 1.10 Calling Class Methods

## What is a Class Method (Static Method)?

A **static method** (also called a **class method**) belongs to the **class itself**, not to any specific object.

**Key Characteristics:**
- Declared with the `static` keyword
- Called using: `ClassName.methodName()`
- **NO need to create an object first**
- **Cannot** access instance variables directly
- Useful for utility functions and shared behavior

---

## Comparison: Static vs Instance Methods

| Feature | Static Method | Instance Method |
|---------|---------------|-----------------|
| **Declaration** | `public static void method()` | `public void method()` |
| **Call syntax** | `ClassName.method()` | `object.method()` |
| **Can access instance variables?** | ❌ No | ✅ Yes |
| **Can access static variables?** | ✅ Yes | ✅ Yes |
| **Needs object to call?** | ❌ No | ✅ Yes |

---

## Example 1: Printer Class

```java
class Printer {
    // STATIC METHOD — Can be called without creating an object
    public static void print(String message) {
        System.out.println(message);
    }

    // INSTANCE METHOD — Needs an object to call
    public void explode() {
        System.out.println("Boom");
    }
}

// Using the methods
Printer.print("Hello!");        // ✅ Works (static method)
// Printer.explode();           // ❌ ERROR (needs object)

Printer p = new Printer();
p.explode();                    // ✅ Works (instance method)
```

## Example 2: Dude Class (Why Static Methods Can't Access Instance Variables)

```java
class Dude {
    String name;  // Instance variable
    
    public Dude(String name) {
        this.name = name;
    }
    
    // ❌ ERROR: Static method can't access instance variable
    public static void sayName() {
        System.out.println("My name is " + name);  // Won't work!
    }
}

// Why? Static methods belong to the class, not to objects.
// They don't know which object's 'name' to use.
```

**Fix:** Either make the method non-static or pass name as a parameter.

```java
// SOLUTION 1: Make it an instance method
public void sayName() {
    System.out.println("My name is " + name);  // ✅ Works
}

// SOLUTION 2: Pass name as parameter
public static void sayName(String name) {
    System.out.println("My name is " + name);  // ✅ Works
}
```

## Critical: Generating Random Integers in a Range

### The Formula

To get a random integer from `min` to `max` (inclusive):


### Step-by-Step Breakdown

```java

// Step 2: Multiply by (6 - 1 + 1) = 6 → 0.0 to 5.999...
// Step 3: Cast to int → 0 to 5
// Step 4: Add min → 1 to 6

System.out.println("You rolled: " + diceRoll);
```

### More Examples


## Quick Reference

---

## 1) Class vs. Object

**Class** = blueprint or template  
**Object** = one specific instance built from that blueprint

### Example

```java
class Dog {
    String name;
    int age;
    
    void bark() {
        System.out.println(name + " says woof!");
    }
}

class Main1 {
    public static void main(String[] args) {
        Dog myDog = new Dog();  // Create object from Dog class
        myDog.name = "Alo";
        myDog.age = 3;
        myDog.bark();           // Alo says woof!
    }
}
```

**Key Point:** The class `Dog` is the blueprint. `myDog` is a specific object (instance) created from that blueprint.

---

## 2) Reference Variables

Variables of reference type store the **memory address** (reference) of an object, not the object itself.

### Key Concept: Multiple References Point to Same Object

```java
Dog d1 = new Dog();
d1.name = "Kai";

Dog d2 = d1;              // Both reference the SAME object
d2.name = "Kairo";        // Change via d2

System.out.println(d1.name);  // Prints "Kairo" (changed!)
```

**Important:** `Dog d2 = d1;` does NOT copy the object. Both variables point to the same Dog in memory.

### Diagram
Changes through either variable affect the shared object.

---

## 3) All Classes Are Subclasses of Object

`Object` is Java's **ultimate superclass**. Every class directly or indirectly extends it.

### What This Means

Every object automatically has these inherited methods:
- `toString()` - returns string representation
- `equals(Object)` - compares objects
- `getClass()` - returns class information
- `hashCode()` - returns hash code

### Example

```java
class Dog {
    String name;
    
    public Dog(String name) {
        this.name = name;
    }
}

class Main3 {
    public static void main(String[] args) {
        Dog x = new Dog("Buddy");
        
        // Methods inherited from Object
        System.out.println(x.getClass().getName());  // Dog
        System.out.println(x.toString());            // Dog@address
        System.out.println(x instanceof Object);     // true
    }
}
```

**Key Point:** All Java objects inherit from `Object`, so they all have these built-in methods available.

## 1.13 Object Creation and Storage (Instantiation)

## Quick Reference

| Concept | Location | Behavior |
|---------|----------|----------|
| **Primitives** (int, double, boolean) | Stack | Stored directly; pass-by-value |
| **Object references** | Stack | Point to heap address |
| **Objects** (Dog, String, arrays) | Heap | Stored in memory; pass-by-reference |

---

## What is Instantiation?

**Instantiation** = creating an actual instance (object) from a class blueprint.

```java
class Dog {
    String name;
    int age;
}

// Instantiation: creating objects from the blueprint
Dog myDog = new Dog();   // Instance 1
Dog yourDog = new Dog(); // Instance 2
```

**Analogy:** Class = cookie cutter, Objects = actual cookies made from it.

---

## Stack vs Heap Memory

### Stack Memory

**Stores:** Primitive data types (int, double, boolean, char, etc.)

**Characteristics:**
- Each thread has its own stack
- Automatically freed when method returns
- Fast allocation
- Limited size (StackOverflowError if exceeded)
- LIFO (Last-In-First-Out) — like stacking plates

```java
public class StackDemo {
    public static void main(String[] args) {
        int number = 100;        // Stack
        double price = 49.99;    // Stack
        boolean isActive = true; // Stack
    }
}
```

**Key Point:** Primitives store their actual value, not a reference.

---

### Heap Memory

**Stores:** Objects and arrays

**Characteristics:**
- Shared among all threads
- Larger than stack
- Slower allocation than stack
- Objects live until no references point to them
- Random access (can grab from anywhere)

```java
public class HeapDemo {
    public static void main(String[] args) {
        String message = new String("Hello");  // Object on heap
        int[] numbers = new int[5];            // Array on heap
        Dog myDog = new Dog();                 // Object on heap
    }
}
```

**Key Point:** Reference variables (on stack) point to objects on heap.

---

## Stack vs Heap Example

```java
public class MemoryDemo {
    public static void main(String[] args) {
        // PRIMITIVES (Stack) — pass-by-value
        int a = 10;
        int b = a;      // Copy of value
        b = 20;
        System.out.println("a = " + a);  // Still 10
        System.out.println("b = " + b);  // Now 20
        
        // OBJECTS (Heap) — pass-by-reference
        int[] array1 = {1, 2, 3};
        int[] array2 = array1;  // Copy of reference (address)
        array2[0] = 99;         // Changes array1 too!
        System.out.println("array1[0] = " + array1[0]);  // 99
        System.out.println("array2[0] = " + array2[0]);  // 99
    }
}
```

**Why?** Primitives copy the value. Objects copy the address (reference), so both point to same object.

---

## Pass-by-Value vs Pass-by-Reference

### Pass-by-Value (Primitives)

When you pass a primitive to a method, Java **copies the value**.
Changes inside the method **do NOT** affect the original.

```java
public class PassByValue {
    public static void changeInt(int n) {
        n = n + 10;  // Only changes the copy
        System.out.println("Inside: n = " + n);  // 15
    }
    
    public static void main(String[] args) {
        int n = 5;
        System.out.println("Before: n = " + n);  // 5
        changeInt(n);
        System.out.println("After: n = " + n);   // Still 5
    }
}
```

**Analogy:** You give your friend a copy of your toy. They paint it red, but your toy stays the same.

---

### Pass-by-Reference (Objects)

When you pass an object to a method, Java **copies the reference** (address).
Changes to the object's **contents** ARE reflected in the original.
But reassigning the reference doesn't affect the original.

```java
class NumberHolder {
    int value;
    NumberHolder(int value) { this.value = value; }
}

public class PassByReference {
    public static void changeValue(NumberHolder n) {
        n.value = n.value + 10;  // Modifies shared object
    }
    
    public static void main(String[] args) {
        NumberHolder myNumber = new NumberHolder(5);
        System.out.println("Before: " + myNumber.value);  // 5
        changeValue(myNumber);
        System.out.println("After: " + myNumber.value);   // 15
    }
}
```

**Analogy:** You give your friend the address of your toy box. They paint a toy inside, so your toys also change (same box).

---

## Popcorn Hack #1: Stack vs Heap

**Question:** Why does changing `b` not affect `a`, but changing `array2` affects `array1`?

**Answer:**
- Primitives (a, b) are stored on stack with their actual values. When you copy `a` to `b`, you copy the value 10. Changing `b` to 20 doesn't affect `a` because they're separate copies.
- Arrays (array1, array2) are stored on heap. The stack variables store references (addresses). When you copy array1 to array2, you copy the address. Both reference the same array object. Changing array2[0] changes the shared object, so array1[0] also changes.

## Key Points & Common Errors

### ✅ DO:

✓ Create an object before calling instance methods  
✓ Call instance methods on the object: `obj.method()`  
✓ Provide arguments matching parameter types  
✓ Use return values from non-void methods  

### ❌ DON'T:

❌ Call instance method on class name: `Dog.bark()` (ERROR)  
❌ Call instance method on null reference (NullPointerException)  
❌ Assign void method return to variable: `int x = obj.doNothing();` (ERROR)  
❌ Use wrong number/type of arguments

## void vs Non-void Methods

### void Methods (Don't Return)

```java
public void bark() {
    System.out.println("Woof!");
}

// Called for side effects only
d.bark();
// Can NOT do: String s = d.bark();  (ERROR)
```

### Non-void Methods (Return Values)

```java
public int getCount() {
    return count;
}

// Must capture or use return value
int val = c.getCount();
System.out.println(c.getCount());
```
