## 3. Peer Review and Design Decisions

I reviewed the analytics design with a peer and used that feedback to refine the graph's organization. One important consideration was that the time divisions should correspond to the class calendar and sprint schedule, rather than being arbitrary date ranges.

I incorporated that direction by separating the graph into sprint-based periods and considering how the same structure could continue to work in future school years. Reviewing the work with a peer helped me evaluate the visualization not only as a graph, but also as a tool that other students and instructors should be able to interpret.

### Design priorities

| Priority | Design decision | Intended benefit |
|---|---|---|
| Clarity | Show commit activity over time | Make project activity easier to interpret |
| Context | Divide the graph by sprint | Connect activity to project phases |
| Consistency | Reference the class calendar | Keep time divisions aligned with the course |
| Sustainability | Avoid a design tied only to one set of dates | Make future-year reuse easier |
| Accessibility | Improve local testing and login access | Make it easier to verify the page during development |
